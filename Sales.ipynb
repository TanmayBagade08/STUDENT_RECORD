# Part 2: Missing Values, Feature Engineering,Filtering and Sorting

```python
 # Check Missing Values
Sales.isnull().sum()
```

```python
 # Rows with Missing Values
Sales[Sales.isnull().any(axis=1)]
```

```python
# Total Missing Values
Sales.isnull().sum().sum()
```

```python
 # Fill Missing Rating with Mean
Sales["Rating"] = Sales["Rating"].fillna(Sales["Rating"].mean())
Sales["Rating"].isnull().sum()
```

```python
# Fill Missing Brand with Mode
Sales["Brand"] = Sales["Brand"].fillna(Sales["Brand"].mode()[0])
Sales["Brand"].isnull().sum() 
```

```python
# Fill Missing Selling Price with Median
Sales["Selling_Price"] = Sales["Selling Price"].fillna(Sales["Selling Price"].median())
```

```python
# Forward Fill
Sales.ffill(inplace=True)
```

```python
# Backward Fill
Sales.bfill(inplace=True)
```

```python
# Remove Missing Rows
Sales.dropna(inplace=True)
```

```python
 # Verify Missing Values
Sales.isnull().sum()
```

 # Feature Engineering

```python
 # Discount Amount
Sales["Discount"] =Sales["Original Price"] - Sales["Selling Price"]
Sales.head(15)
```

```python
# Discount Percentage
Sales["Discount_Percentage"] = ( (Sales["Original Price"] - Sales["Selling Price"]) / Sales["Original Price"]) * 100

Sales.head(10)
```

```python
 # Premium or Budget Phone

Sales["Phone_Category"] = np.where( Sales["Selling_Price"] > 50000, "Premium", "Budget")
Sales.head(15)
```

```python
 # Rating Category
Sales["Rating_Category"] = np.where(Sales["Rating"] >= 4.5, "Excellent",np.where(Sales["Rating"] >= 4.0,"Good","Average"))
Sales.head(15)
```

```python
# Price Difference
Sales["Price Difference"] =Sales["Original Price"] -Sales["Selling Price"]
```
