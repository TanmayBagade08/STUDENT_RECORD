```python
import os

FILE = "students.txt"
 
def load_data():
    students = []
    if os.path.exists(FILE):
        with open(FILE, "r") as f:
            for line in f:
                name, roll, marks = line.strip().split(",")
                students.append({
                    "name": name,
                    "roll": int(roll),
                    "marks": float(marks)
                })
    return students
 
def save_data(students):
    with open(FILE, "w") as f:
        for s in students:
            f.write(f"{s['name']},{s['roll']},{s['marks']}\n")
 
def add_student():
    students = load_data()

    name = input("Enter Name: ")
    roll = int(input("Enter Roll Number: "))
    marks = float(input("Enter Marks: "))

    students.append({
        "name": name,
        "roll": roll,
        "marks": marks
    })

    save_data(students)
    print("Student Added Successfully!\n")
 
def view_students():
    students = load_data()

    if not students:
        print("No Records Found.\n")
        return

    print("\nName\tRoll\tMarks")
    print("-" * 30)

    for s in students:
        print(f"{s['name']}\t{s['roll']}\t{s['marks']}")
 
def search_student():
    students = load_data()

    roll = int(input("Enter Roll Number: "))

    for s in students:
        if s["roll"] == roll:
            print("\nStudent Found")
            print("Name :", s["name"])
            print("Marks:", s["marks"])
            return

    print("Student Not Found.")

 
def update_marks():
    students = load_data()

    roll = int(input("Enter Roll Number: "))

    for s in students:
        if s["roll"] == roll:
            s["marks"] = float(input("Enter New Marks: "))
            save_data(students)
            print("Marks Updated Successfully.")
            return

    print("Student Not Found.")


def delete_student():
    students = load_data()

    print("1. Delete by Name")
    print("2. Delete by Roll")

    choice = input("Enter Choice: ")

    if choice == "1":
        name = input("Enter Name: ")
        students = [s for s in students if s["name"] != name]

    elif choice == "2":
        roll = int(input("Enter Roll Number: "))
        students = [s for s in students if s["roll"] != roll]

    else:
        print("Invalid Choice")
        return

    save_data(students)
    print("Record Deleted.")
 
def sort_by_marks():
    students = load_data()

    students.sort(key=lambda x: x["marks"], reverse=True)

    print("\nStudents Sorted by Marks")
    print("-" * 30)

    for s in students:
        print(s["name"], s["roll"], s["marks"])

 
def above_75():
    students = load_data()

    print("\nStudents with Marks > 75")

    for s in students:
        if s["marks"] > 75:
            print(s["name"], s["roll"], s["marks"])

def export_top3():
    students = load_data()

    students.sort(key=lambda x: x["marks"], reverse=True)

    with open("top3.txt", "w") as f:
        for s in students[:3]:
            f.write(f"{s['name']},{s['roll']},{s['marks']}\n")

    print("Top 3 Students Exported to top3.txt")
 
def backup():
    with open(FILE, "r") as f1, open("backup.txt", "w") as f2:
        f2.write(f1.read())

    print("Backup Created Successfully.")
 
def count_students():
    students = load_data()

    above90 = 0
    below35 = 0

    for s in students:
        if s["marks"] > 90:
            above90 += 1
        elif s["marks"] < 35:
            below35 += 1

    print("Students scoring > 90 :", above90)
    print("Students scoring < 35 :", below35)
while True:

    print("\n===== STUDENT RECORD SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Delete Student")
    print("6. Sort by Marks")
    print("7. Display Marks > 75")
    print("8. Export Top 3")
    print("9. Backup Data")
    print("10. Count (>90 and <35)")
    print("11. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_student()

    elif choice == "2":
        view_students()

    elif choice == "3":
        search_student()

    elif choice == "4":
        update_marks()

    elif choice == "5":
        delete_student()

    elif choice == "6":
        sort_by_marks()

    elif choice == "7":
        above_75()

    elif choice == "8":
        export_top3()

    elif choice == "9":
        backup()

    elif choice == "10":
        count_students()

    elif choice == "11":
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
